### Task 1 — Grocery Basket



```python
# Start with an empty shopping list (list).
# 1. Append at least 4 items supplied in one line of user input (comma-separated).
# 2. Convert the list to a *tuple* called immutable_basket.
# 3. Print the third item using tuple indexing.

# Start with an empty shopping list
shopping_list = []

# Get user input (comma-separated)
items_input = input("Enter at least 4 grocery items (comma-separated): ")

# Split input and append items to the list
items = items_input.split(",")

for item in items:
    shopping_list.append(item.strip())

# Convert list to tuple
immutable_basket = tuple(shopping_list)

# Print the third item
print("Third item:", immutable_basket[2])
```

### Task 2 — Word Stats

```python
sample = "to be or not to be that is the question"

# 1. Build a set `unique_words` containing every distinct word.
# 2. Build a dict `word_counts` mapping each word to the number of times it appears.
#    (Hint: .split() + a simple loop)
# 3. Print the two structures and explain (in a comment) their main difference.

sample = "to be or not to be that is the question"

# Split the sentence into words
words = sample.split()

# 1. Build a set of unique words
unique_words = set(words)

# 2. Build a dictionary of word counts
word_counts = {}

for word in words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# 3. Print the results
print("Unique words:", unique_words)
print("Word counts:", word_counts)

# Difference:
# - set stores only unique words (no duplicates, no counts)
# - dict stores each word with the number of times it appears

```

 Task 1 — Prime Tester

```python
def is_prime(n: int) -> bool:
    """
    Return True if n is a prime number, else False.
    0 and 1 are *not* prime.
    """
    def is_prime(n: int) -> bool:
    """
    Return True if n is a prime number, else False.
    0 and 1 are *not* prime.
    """
    if n < 2:
        return False  # 0 و 1 عدد اول نیستند

    for i in range(2, n):
        if n % i == 0:
            return False  # اگر بخش‌پذیر بود → اول نیست

    return True  # اگر هیچ بخش‌پذیری پیدا نشد → عدد اول

# Quick self-check
print([x for x in range(10) if is_prime(x)])   # Expected: [2, 3, 5, 7]
```

### Task 2 — Repeater Greeter

```python
def greet(name: str, times: int = 1) -> None:
    """Print `name`, capitalised, exactly `times` times on one line."""
    # Capitalize the name
    cap_name = name.capitalize()

    # Repeat the name 'times' times and join with space
    print(" ".join([cap_name] * times))


# Test examples
greet("alice")          # Alice
greet("bob", times=3)   # Bob Bob Bob


greet("alice")          # Alice
greet("bob", times=3)   # Bob Bob Bob
```

### Task 1 — Simple Counter

```python
class Counter:
    """Counts how many times `increment` is called."""

    def __init__(self):
        # 1️⃣ متغیر داخلی شمارنده، شروع از 0
        self.count = 0

    def increment(self, step: int = 1):
        # 2️⃣ افزودن step به شمارنده
        self.count += step

    def value(self):
        # 3️⃣ برگرداندن مقدار فعلی شمارنده
        return self.count



c = Counter()

for _ in range(5):
    c.increment()  # هر بار 1 واحد افزایش

print(c.value())   # خروجی مورد انتظار: 5
```

### Task 2 — 2-D Point with Distance

```python
import math

class Point:
    """
    A 2-D point supporting distance calculation.
    Usage:
        p = Point(3, 4)
        q = Point(0, 0)
        print(p.distance_to(q))  # 5.0
    """

    def __init__(self, x, y):
        # 1️⃣ ذخیره مختصات x و y
        self.x = x
        self.y = y

    def distance_to(self, other):
        # 2️⃣ محاسبه فاصله اقلیدسی
        dx = self.x - other.x
        dy = self.y - other.y
        return math.sqrt(dx**2 + dy**2)


# Smoke test
p, q = Point(3, 4), Point(0, 0)
assert round(p.distance_to(q), 1) == 5.0
print(p.distance_to(q))  # 5.0
```
